```python

import math

# Catálogo de películas
peliculas = {
    "Titanic": ["romance", "drama", "barco", "amor"],
    "Avatar": ["ciencia ficcion", "aventura", "accion", "espacio"],
    "Matrix": ["ciencia ficcion", "accion", "tecnologia"],
    "Interstellar": ["ciencia ficcion", "espacio", "drama"],
    "Inception": ["ciencia ficcion", "accion", "misterio"],
    "Gladiator": ["accion", "aventura", "drama", "guerra"],
    "Toy Story": ["animacion", "comedia", "aventura"],
    "Frozen": ["animacion", "aventura", "fantasia"],
    "Shrek": ["animacion", "comedia", "fantasia"],
    "Coco": ["animacion", "familia", "musica", "drama"],
    "Joker": ["drama", "crimen", "misterio"],
    "Batman": ["accion", "crimen", "misterio"]
}

# Crear vocabulario
vocabulario = sorted(set(
    palabra for palabras in peliculas.values()
    for palabra in palabras
))

# Convertir palabras clave en un vector
def crear_vector(palabras):
    return [int(p in palabras) for p in vocabulario]

# Calcular similitud del coseno
def similitud(v1, v2):
    producto = sum(a * b for a, b in zip(v1, v2))
    norma1 = math.sqrt(sum(a ** 2 for a in v1))
    norma2 = math.sqrt(sum(b ** 2 for b in v2))

    if norma1 == 0 or norma2 == 0:
        return 0

    return producto / (norma1 * norma2)

# Función 1: comparar dos películas
def comparar_peliculas():
    print("\nPelículas disponibles:")
    for titulo in peliculas:
        print("-", titulo)

    p1 = input("\nPrimera película: ").strip().lower()
    p2 = input("Segunda película: ").strip().lower()

    nombres = {titulo.lower(): titulo for titulo in peliculas}

    if p1 not in nombres or p2 not in nombres:
        print("Película no encontrada")
        return

    p1, p2 = nombres[p1], nombres[p2]

    v1 = crear_vector(peliculas[p1])
    v2 = crear_vector(peliculas[p2])

    resultado = similitud(v1, v2)
    print(f"\nSimilitud entre {p1} y {p2}: {resultado:.2%}")

# Función 2: buscar películas por palabras clave
def buscar_peliculas():
    entrada = input("\nPalabras clave separadas por comas: ")
    palabras = [p.strip().lower() for p in entrada.split(",")]

    vector_busqueda = crear_vector(palabras)
    resultados = []

    for titulo, etiquetas in peliculas.items():
        vector = crear_vector(etiquetas)
        valor = similitud(vector_busqueda, vector)

        if valor > 0:
            resultados.append((titulo, valor))

    resultados.sort(key=lambda x: x[1], reverse=True)

    print("\nPelículas encontradas:")
    if not resultados:
        print("No hay coincidencias")

    for titulo, valor in resultados:
        print(f"{titulo}: {valor:.2%}")


# Menú principal
def main():
    while True:
        print("\n1. Comparar dos películas")
        print("2. Buscar películas por palabras clave")
        print("3. Salir")

        opcion = input("Selecciona una opción: ")

        if opcion == "1":
            comparar_peliculas()
        elif opcion == "2":
            buscar_peliculas()
        elif opcion == "3":
            break
        else:
            print("Opción no válida")


```
